
# INTRODUCTION


# PART 1 — INTELLIGENCE KYA HAI?

## Definition
Intelligence wo ability hai jisme insaan ya machine seekh sakta hai, samajh sakta hai, decision le sakta hai, aur naye situations mein adjust kar sakta hai.

## Simple Explanation 
Socho ek student hai. Wo exam se pehle kitaab padhta hai, notes banata hai, formulas yaad karta hai. Ye hai **learning**. Phir wo exam hall mein baithta hai aur sawal dekhta hai. Usay samajh aata hai ke ye question kis type ka hai. Ye hai **pattern recognition**. Phir wo decide karta hai ke pehle kaunsa question solve karna hai. Ye hai **decision making**. Agar question mushkil ho to wo tootta nahi, balki apne dimaag ko lagata hai. Ye hai **problem solving**. Agar exam pattern change ho jaye to wo adjust karta hai. Ye hai **adaptation**. In sab cheezon ka naam hi intelligence hai.

Intelligence sirf maths ya science nahi hai. Intelligence mein **memory, reasoning, planning, aur seekhne ki ability** sab shamil hain. Insaan ka dimaag **86 billion neurons** se bana hai. Jab hum kuch seekhte hain to neurons ke beech connections strong hote hain. Jab bhoolte hain to connections weak ho jate hain. AI bhi isi tarah kaam karta hai, lekin wo **biological nahi, mathematical** hai.

## Real-World Example
Ek doctor patient ko dekhta hai. Usay symptoms sun kar samajh aata hai ke ye bimari kya ho sakti hai. Ye intelligence hai. Ek cricketer ball ko dekh kar samajh jata hai ke spin hai ya pace. Ye bhi intelligence hai.

## How It Works (Behind the Scenes)
Intelligence ke andar ye components kaam karte hain:

| Component | Roman Urdu Explanation |
|---|---|
| Learning | Naya information lena aur yaad rakhna |
| Pattern Recognition | Cheezon mein similarity dhoondna |
| Decision Making | Options mein se best choose karna |
| Problem Solving | Mushkil ka hal nikalna |
| Adaptation | Naye situation ke mutabiq khud ko change karna |

```
┌─────────────────────────────────────────────────────────┐
│                    INTELLIGENCE                          │
│                                                          │
│   ┌──────────┐  ┌──────────┐  ┌──────────┐              │
│   │ LEARNING │  │ PATTERN  │  │ DECISION │              │
│   │          │  │RECOGNITION│  │ MAKING   │              │
│   └────┬─────┘  └────┬─────┘  └────┬─────┘              │
│        │              │              │                    │
│        └──────────────┼──────────────┘                    │
│                       │                                   │
│              ┌────────▼────────┐                         │
│              │  PROBLEM SOLVING │                         │
│              └────────┬────────┘                         │
│                       │                                   │
│              ┌────────▼────────┐                         │
│              │   ADAPTATION    │                         │
│              └─────────────────┘                         │
│                                                          │
│   Human Brain: 86 Billion Neurons                        │
│   AI: Mathematical Neurons (Not Biological)              │
└─────────────────────────────────────────────────────────┘
```

## Key Points
- Intelligence sirf maths nahi — memory, reasoning, planning, learning sab isme aate hain
- Insaan ka dimaag 86 billion neurons se bana hai
- AI bhi neurons se inspired hai lekin mathematical hai, biological nahi
- Learning se connections strong hote hain, bhoolne se weak

# PART 2 — ARTIFICIAL INTELLIGENCE KYA HAI?

## Definition
Artificial Intelligence ka matlab hai machine ko aisa smart banana ke wo aise kaam kar sake jo normally insaan karta hai.

## Simple Explanation 
AI insaan nahi hai. AI ko **consciousness nahi hoti**. AI ko **emotions nahi hote**. AI sirf **data se patterns seekhta hai**. Google Maps traffic predict karta hai, YouTube aapko videos suggest karta hai, face recognition aapka chehra pehchaan kar phone unlock karta hai, chatbot aapke sawalon ka jawab deta hai, bank fraud pakadta hai, doctor ko X-ray mein disease detect karne mein madad karta hai, aur self-driving car road par safely drive karti hai. Ye sab AI hai.

AI ka purpose hai:
- Mushkil kaam asaan karna
- Repetitive kaam automate karna
- Bade data se patterns nikalna
- Predictions karna

## Traditional Programming vs AI — Key Difference

| Traditional Programming | Artificial Intelligence |
|---|---|
| Insaan rules likhta hai | Insaan data deta hai |
| Computer rules follow karta hai | Machine khud rules seekhta hai |
| Deterministic (fixed output) | Probabilistic (probability-based) |
| Naye situation ke liye naya rule likhna padta hai | Model generalize kar sakta hai |

## Visual Diagram — AI vs Traditional Programming

## Real-World Example
**Traditional:** Agar aap chahte hain ke computer bataye ke student pass hoga ya fail, to aap rule likhenge: "Agar marks 40 se zyada to pass." Lekin AI mein aap 1000 students ka data denge aur machine khud seekhegi ke kaunse factors pass/fail se related hain.

## Key Points
- AI insaan nahi — consciousness, emotions nahi hoti
- AI data se patterns seekhta hai
- Traditional programming mein rules insaan likhta hai, AI mein machine khud seekhti hai
- AI ka purpose: mushkil kaam asaan, repetitive automate, patterns nikalna, predictions

AI ka safar 1950 se shuru hota hai aur aaj tak continue hai. Har era ne pichhle era ki limitation ko door kiya.

**2017 — Transformer Paper:** "Attention Is All You Need" aaya.

**2020 — GPT-3:** 175 billion parameters.

**2022 — ChatGPT:** Launch hua aur duniya badal gayi.

**2023 — GPT-4, Claude, Gemini:** Advanced models aaye.

**2024 — AI Agents:** Autonomously kaam karne wale agents ka daur.

# PART 5 — TRADITIONAL PROGRAMMING vs AI

## Definition
Traditional programming mein rules insaan likhta hai. AI mein machine data se rules seekhti hai.

## Simple Explanation 
Traditional programming mein aap rules likhte hain aur computer output deta hai. Jaise: "Agar age 18 se zyada hai to adult." Machine Learning mein aap data aur examples dete hain, machine khud rules seekhta hai. Jaise aap 1000 students ka data dete hain jisme study hours, attendance, previous marks aur final marks hain. Machine seekh jati hai ke study hours aur attendance ka final marks se kya relation hai.

Traditional programming **deterministic** hai — same input, same output. ML **probabilistic** hai — same input, different probability. Traditional programming mein naye situation ke liye naya rule likhna padta hai. ML model **generalize** kar sakta hai.

Traditional programming **recipe book** ki tarah hai. ML **chef training** ki tarah hai — chef ko recipes yaad nahi, wo seekh gaya hai ke ingredients kaise combine karne hain.

## Visual Diagram — Traditional vs ML

```
TRADITIONAL PROGRAMMING:
┌─────────────┐     ┌─────────────┐     ┌─────────────┐
│   INPUT     │────▶│   RULES     │────▶│   OUTPUT    │
│  (Age: 20)  │     │ (If age>18  │     │  (Adult)    │
│             │     │  = Adult)   │     │             │
└─────────────┘     └─────────────┘     └─────────────┘
                    (Insaan ne likhe)

MACHINE LEARNING:
┌─────────────┐     ┌─────────────┐     ┌─────────────┐
│   DATA      │────▶│   MODEL     │────▶│   RULES     │
│ (1000       │     │ (Learns     │     │ (Khud       │
│  students)  │     │  patterns)  │     │  seekhe)    │
└─────────────┘     └─────────────┘     └─────────────┘
                                              │
                                              ▼
┌─────────────┐     ┌─────────────┐     ┌─────────────┐
│   INPUT     │────▶│   MODEL     │────▶│   OUTPUT    │
│  (Age: 20)  │     │ (Applies    │     │  (Adult)    │
│             │     │  learned    │     │             │
│             │     │  rules)     │     │             │
└─────────────┘     └─────────────┘     └─────────────┘
```

## Real-World Example
**Traditional:** Email filter banana — aap rule likhenge "Agar email mein 'lottery' word hai to spam." Lekin spammer 'l0ttery' likh dega to rule fail.

**ML:** Aap 10,000 emails ka data denge jisme spam aur not spam labeled hain. Machine khud seekhegi ke kaunse patterns spam indicate karte hain.

## Key Points
- Traditional: Rules → Output (insaan rules likhta)
- ML: Data → Rules → Output (machine rules seekhti)
- Traditional deterministic, ML probabilistic
- Traditional recipe book, ML chef training

# PART 6 — MACHINE LEARNING

## Definition
Machine Learning wo tarika hai jisme machine data se khud seekhti hai bina explicitly program kiye.

## Simple Explanation 
Machine Learning ki kuch types hoti hain:

**1. Supervised Learning:** Aap input aur correct output dono dete hain. Jaise teacher student ko question aur answer dono dikhata hai. Isme:
- **Classification:** Output category hoti hai (spam/not spam, cat/dog)
- **Regression:** Output number hota hai (house price, temperature)

**2. Unsupervised Learning:** Aap sirf input dete hain, output nahi. Machine khud patterns dhoondta hai. Isme:
- **Clustering:** Customer segmentation
- **Pattern Discovery:** Market basket analysis
- **Dimensionality Reduction:** 1000 features ko 50 features mein convert karna

**3. Semi-Supervised Learning:** Thoda labeled data aur bohat sara unlabeled data.

**4. Self-Supervised Learning:** Data khud labels generate karta hai, jaise LLM training.

**5. Reinforcement Learning:** Agent environment mein actions leta hai, rewards milte hain, aur wo seekhta hai. Isme Agent, Environment, State, Action, Reward, aur Policy hote hain. Jaise puppy ko train karna — acha kaam pe treat, bura kaam pe no.

## Visual Diagram — ML Learning Types

```
┌─────────────────────────────────────────────────────────────────┐
│                    MACHINE LEARNING TYPES                        │
├─────────────────────────────────────────────────────────────────┤
│                                                                  │
│  ┌──────────────────┐  ┌──────────────────┐  ┌───────────────┐ │
│  │   SUPERVISED     │  │  UNSUPERVISED    │  │ REINFORCEMENT │ │
│  │                  │  │                  │  │               │ │
│  │ Input + Output   │  │ Sirf Input       │  │ Agent +       │ │
│  │ dono dete hain   │  │ dete hain        │  │ Environment   │ │
│  │                  │  │                  │  │               │ │
│  │ ┌──────────────┐ │  │ ┌──────────────┐ │  │ ┌───────────┐ │ │
│  │ │Classification│ │  │ │ Clustering   │ │  │ │  Agent    │ │ │
│  │ │(Spam/Not Spam)│ │  │ │(Segmentation)│ │  │ │    ↓      │ │ │
│  │ └──────────────┘ │  │ └──────────────┘ │  │ │  Action   │ │ │
│  │ ┌──────────────┐ │  │ ┌──────────────┐ │  │ │    ↓      │ │ │
│  │ │ Regression   │ │  │ │  Pattern     │ │  │ │  Reward   │ │ │
│  │ │(House Price) │ │  │ │  Discovery   │ │  │ │    ↓      │ │ │
│  │ └──────────────┘ │  │ └──────────────┘ │  │ │  Learn    │ │ │
│  └──────────────────┘  └──────────────────┘  │ └───────────┘ │ │
│                                               └───────────────┘ │
│                                                                  │
│  ┌──────────────────┐  ┌──────────────────┐                     │
│  │ SEMI-SUPERVISED  │  │ SELF-SUPERVISED  │                     │
│  │                  │  │                  │                     │
│  │ Thoda labeled +  │  │ Data khud labels │                     │
│  │ Bohat unlabeled  │  │ generate karta   │                     │
│  │                  │  │ (LLM Training)   │                     │
│  └──────────────────┘  └──────────────────┘                     │
└─────────────────────────────────────────────────────────────────┘
```

## Real-World Example
- **Supervised:** Doctor X-ray dekh kar batata hai "ye pneumonia hai" — machine seekhti hai
- **Unsupervised:** Netflix dekhta hai ke kaunse users similar hain, unhe similar movies suggest karta hai
- **Reinforcement:** Robot ko chalna sikhana — girta hai to negative reward, chalta hai to positive

## Key Points
- Supervised: Input + Output dono
- Unsupervised: Sirf Input, patterns khud dhoondta
- Semi-Supervised: Thoda labeled + bohat unlabeled
- Self-Supervised: Data khud labels banata
- Reinforcement: Agent + Environment + Reward

# PART 8 — DATA QUALITY

## Definition
Data quality bohat important hai kyunki ganda data = ganda model.

## Simple Explanation 
**Missing Data:** Jab kuch values available nahi hoti.

**Duplicate Data:** Jab ek hi record do baar aa jata hai.

**Incorrect Data:** Jab galat values hoti hain jaise age -5.

**Noisy Data:** Random errors.

**Outliers:** Wo values jo baaki data se bohat door hoti hain. Jaise 45, 50, 52, 49, 51, 48, **99** mein 99 outlier hai.

Lekin har outlier galat nahi hota:
- Fraud detection mein outlier fraud ka signal ho sakta hai
- Medical data mein outlier bimari ka indication ho sakta hai
- Sensor data mein outlier error ho sakta hai

**Mean, Median, IQR, aur Z-score** outlier detect karne ke tarike hain.

**Data Bias:** Jab data kisi ek group ki taraf jhuka ho.

**Data Leakage:** Jab training data mein test data ki information aa jaye.

## Visual Diagram — Data Quality Issues

```
┌─────────────────────────────────────────────────────────────────┐
│                    DATA QUALITY PROBLEMS                         │
├─────────────────────────────────────────────────────────────────┤
│                                                                  │
│  MISSING DATA:                                                   │
│  ┌──────┬──────┬──────┬──────┐                                  │
│  │  45  │  ?   │  52  │  48  │  ← Missing value                │
│  └──────┴──────┴──────┴──────┘                                  │
│                                                                  │
│  DUPLICATE DATA:                                                 │
│  ┌──────┬──────┐                                                │
│  │  45  │  45  │  ← Same record twice                           │
│  └──────┴──────┘                                                │
│                                                                  │
│  INCORRECT DATA:                                                 │
│  ┌──────┐                                                       │
│  │ -5   │  ← Age can't be negative                              │
│  └──────┘                                                       │
│                                                                  │
│  OUTLIER:                                                        │
│  ┌───┬───┬───┬───┬───┬───┬─────┐                                │
│  │45 │50 │52 │49 │51 │48 │ 99  │  ← Outlier (baaki se door)     │
│  └───┴───┴───┴───┴───┴───┴─────┘                                │
│                                                                  │
│  DATA BIAS:                                                      │
│  ┌─────────────────────────────────────┐                        │
│  │ 90% data ek group ka, 10% doosre ka │  ← Biased              │
│  └─────────────────────────────────────┘                        │
│                                                                  │
│  DATA LEAKAGE:                                                   │
│  ┌─────────────────────────────────────┐                        │
│  │ Training data mein test data ki info │  ← Cheating            │
│  └─────────────────────────────────────┘                        │
└─────────────────────────────────────────────────────────────────┘
```

## Real-World Example
Agar aap ek app bana rahe hain jo diabetes predict kare, aur aapka data sirf 20-30 saal ke logon ka hai, to 60 saal ke patient par model fail hoga. Ye **data bias** hai.

## Key Points
- Missing, Duplicate, Incorrect, Noisy, Outliers — sab data quality issues
- Har outlier galat nahi — fraud/medical mein important

# PART 17 — SELF-ATTENTION

**Query, Key, Value** se **Attention Score** banta hai. Phir **Weighted Information** milti hai. Phir **Updated Representation** banti hai.

## Visual Diagram — Self-Attention

```
┌─────────────────────────────────────────────────────────────────┐
│                      SELF-ATTENTION                              │
│                                                                  │
│  SENTENCE: "The cat sat on the mat because it was tired."       │
│                                                                  │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  STEP 1: Query, Key, Value nikaalna                     │    │
│  │                                                          │    │
│  │  "it" ──▶ Query (Q)                                     │    │
│  │  "cat" ──▶ Key (K)                                      │    │
│  │  "mat" ──▶ Key (K)                                      │    │
│  └─────────────────────────────────────────────────────────┘    │
│                                                                  │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  STEP 2: Attention Score = Q × K                        │    │
│  │                                                          │    │
│  │  "it" vs "cat" ──▶ Score: 0.9 (HIGH)  ← Related         │    │
│  │  "it" vs "mat" ──▶ Score: 0.1 (LOW)   ← Not related     │    │
│  └─────────────────────────────────────────────────────────┘    │
│                                                                  │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  STEP 3: Softmax Apply                                  │    │
│  │                                                          │    │
│  │  "cat" ──▶ 0.9 ──▶ Softmax ──▶ 0.85 (85% attention)    │    │
│  │  "mat" ──▶ 0.1 ──▶ Softmax ──▶ 0.15 (15% attention)    │    │
│  └─────────────────────────────────────────────────────────┘    │
│                                                                  │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  STEP 4: Weighted Information                           │    │
│  │                                                          │    │
│  │  0.85 × "cat" Value + 0.15 × "mat" Value                │    │
│  │  = Updated Representation of "it"                       │    │
│  │  (Ab "it" ka matlab cat ke qareeb hai)                  │    │
│  └─────────────────────────────────────────────────────────┘    │
└─────────────────────────────────────────────────────────────────┘
```

## Real-World Example
Jaise aap ek group photo dekhte hain aur poochte hain "ye kaun hai?" Aap foran us bande ko pehchante hain jo aapke dost ke paas khada hai. Self-attention bhi aise hi words ke relationships samajhta hai.

## Key Points
- Self-Attention: Har word doosre se related
- Query, Key, Value
- Attention Score = Q × K
- Softmax: Normalize
- Weighted Information: Updated representation

# PART 21 — DECODING

## Definition
Decoding wo process hai jisme model probabilities se token choose karta hai.

## Simple Explanation 
**Greedy Decoding:** Sabse zyada probability wala token choose hota hai.
**Sampling:** Random choose hota hai.
**Temperature:** Batata hai ke randomness kitni ho.
- Low temperature: Model safe hota hai
- High temperature: Model creative hota hai
**Top-K:** Top K tokens mein se choose hota hai.
**Top-P:** Cumulative probability se choose hota hai.

## Visual Diagram — Decoding Strategies

```
┌─────────────────────────────────────────────────────────────────┐
│                    DECODING STRATEGIES                           │
│                                                                  │
│  PROBABILITIES:                                                  │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  Token A: 0.40                                          │    │
│  │  Token B: 0.30                                          │    │
│  │  Token C: 0.15                                          │    │
│  │  Token D: 0.10                                          │    │
│  │  Token E: 0.05                                          │    │
│  └─────────────────────────────────────────────────────────┘    │
│                                                                  │
│  GREEDY DECODING:                                                │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  Always choose Token A (0.40)                           │    │
│  │  Result: Predictable, safe                              │    │
│  └─────────────────────────────────────────────────────────┘    │
│                                                                  │
│  SAMPLING:                                                       │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  Randomly choose based on probability                   │    │
│  │  Result: Creative, unpredictable                        │    │
│  └─────────────────────────────────────────────────────────┘    │
│                                                                  │
│  TEMPERATURE:                                                    │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  Low (0.1):  ████████████████████  Token A (safe)      │    │
│  │  High (1.5): ████████  ██████  ████  ██  (creative)    │    │
│  └─────────────────────────────────────────────────────────┘    │
│                                                                  │
│  TOP-K:                                                          │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  Sirf Top 3 tokens mein se choose                       │    │
│  │  A, B, C mein se koi bhi                                │    │
│  └─────────────────────────────────────────────────────────┘    │
│                                                                  │
│  TOP-P:                                                          │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  Cumulative probability 0.90 tak                    │    │
│  │  A (0.40) + B (0.30) + C (0.15) + D (0.05) = 0.90     │    │
│  │  In 4 mein se choose                                    │    │
│  └─────────────────────────────────────────────────────────┘    │
└─────────────────────────────────────────────────────────────────┘
```

## Real-World Example
Jaise aap ice cream choose karte hain:
- **Greedy:** Hamesha chocolate (favourite)
- **Sampling:** Kabhi chocolate, kabhi vanilla, kabhi strawberry
- **Temperature:** Low = hamesha same, High = har baar naya
- **Top-K:** Sirf top 3 flavours mein se
- **Top-P:** 90% probability tak ke flavours mein se

## Key Points
- Greedy: Highest probability
- Sampling: Random
- Temperature: Randomness control
- Top-K: Top K tokens
- Top-P: Cumulative probability

## Definition
Chain-of-Thought ek technique hai jisme model ko step by step sochne ke liye kaha jata hai.

## Simple Explanation 
Jaise pehle ye karo, phir ye, phir ye. Isse complex problems solve karna asaan ho jata hai. Lekin **private chain-of-thought** ko expose nahi karna chahiye, sirf **high-level reasoning** explain karni chahiye.

```
┌─────────────────────────────────────────────────────────────────┐
│                    CHAIN OF THOUGHT                              │
│                                                                  │
│  WITHOUT CHAIN OF THOUGHT:                                       │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  Q: "Agar 3 apples 15 rupees, to 7 apples?"             │    │
│  │  A: "35 rupees"  ← Direct answer, reasoning nahi        │    │
│  └─────────────────────────────────────────────────────────┘    │
│                                                                  │
│  WITH CHAIN OF THOUGHT:                                          │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  Q: "Agar 3 apples 15 rupees, to 7 apples?"             │    │
│  │                                                          │    │
│  │  Step 1: Pehle 1 apple ki price nikaalte hain          │    │
│  │          15 ÷ 3 = 5 rupees                               │    │
│  │                                                          │    │
│  │  Step 2: Ab 7 apples ki price nikaalte hain            │    │
│  │          5 × 7 = 35 rupees                               │    │
│  │                                                          │    │
│  │  Step 3: Check karte hain                               │    │
│  │          35 ÷ 7 = 5 ✓                                    │    │
│  │                                                          │    │
│  │  A: "35 rupees"                                          │    │
│  └─────────────────────────────────────────────────────────┘    │
│                                                                  │
│  PRIVATE CHAIN OF THOUGHT:                                       │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  Model ke internal steps ko expose nahi karna           │    │
│  │  Sirf high-level reasoning explain karni                │    │
│  └─────────────────────────────────────────────────────────┘    │
└─────────────────────────────────────────────────────────────────┘
```

# PART 26 — HALLUCINATION

## Definition
Hallucination tab hota hai jab AI galat ya made-up information deta hai.

## Simple Explanation

# PART 27 — RAG (Retrieval-Augmented Generation)

## Definition
RAG ka matlab hai Retrieval-Augmented Generation. Isme model ko external knowledge di jati hai.

## Simple Explanation 
**Without RAG:** Question → LLM → Answer
**With RAG:** Question → Search → Relevant Documents → Context → LLM → Answer

Isme use hote hain:
- **Retrieval:** Relevant documents dhoondna
- **Chunking:** Documents ko chhote hisson mein todna
- **Embeddings:** Chunks ko vectors mein convert
- **Vector Database:** Embeddings store karna
- **Similarity Search:** Similar chunks dhoondna
- **Reranking:** Best chunks ko rank karna

Jaise aap PDF chatbot banate hain. Pehle PDF ke chunks banate hain, unke embeddings store karte hain, phir user ke question ke liye relevant chunks dhoondte hain, aur unhe LLM ko dete hain.

## Visual Diagram — RAG Architecture

```
┌─────────────────────────────────────────────────────────────────┐
│                    RAG ARCHITECTURE                              │
│                                                                  │
│  WITHOUT RAG:                                                    │
│  ┌──────────┐    ┌──────────┐    ┌──────────┐                   │
│  │ QUESTION │───▶│   LLM    │───▶│  ANSWER  │                   │
│  └──────────┘    └──────────┘    └──────────┘                   │
│                  (Sirf apni knowledge)                           │
│                                                                  │
│  WITH RAG:                                                       │
│  ┌──────────┐    ┌──────────┐    ┌──────────┐                   │
│  │ QUESTION │───▶│  SEARCH  │───▶│ RELEVANT │                   │
│  │          │    │          │    │ DOCUMENTS│                   │
│  └──────────┘    └──────────┘    └────┬─────┘                   │
│                                        │                         │
│                                        ▼                         │
│  ┌──────────┐    ┌──────────┐    ┌──────────┐                   │
│  │  ANSWER  │◀───│   LLM    │◀───│ CONTEXT  │                   │
│  │          │    │          │    │ + QUESTION│                   │
│  └──────────┘    └──────────┘    └──────────┘                   │
│                                                                  │
│  COMPLETE RAG PIPELINE:                                          │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  DOCUMENTS ──▶ CHUNKING ──▶ EMBEDDINGS ──▶ VECTOR DB   │    │
│  │                                                          │    │
│  │  USER QUESTION ──▶ EMBEDDING ──▶ SIMILARITY SEARCH      │    │
│  │                                                          │    │
│  │  RELEVANT CHUNKS ──▶ RERANKING ──▶ CONTEXT ──▶ LLM      │    │
│  │                                                          │    │
│  │  LLM ──▶ ANSWER (Context ke saath)                      │    │
│  └─────────────────────────────────────────────────────────┘    │
└─────────────────────────────────────────────────────────────────┘
```

## Real-World Example
Jaise aap exam mein jaate hain. Aapko sab kuch yaad nahi, lekin aap ek **cheat sheet** le jate hain. Jab question aata hai, aap cheat sheet mein relevant information dhoondte hain aur uske hisaab se answer likhte hain. RAG bhi aisa hi hai.

## Key Points
- RAG = Retrieval + Generation
- Without RAG: LLM apni knowledge se
- With RAG: External documents se
- Retrieval, Chunking, Embeddings, Vector DB, Similarity Search, Reranking
- PDF chatbot ka example

```
┌─────────────────────────────────────────────────────────────────┐
│           PROMPTING vs RAG vs FINE-TUNING                       │
│                                                                  │
│  PROMPTING:                                                      │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  User ──▶ Prompt ──▶ LLM ──▶ Output                     │    │
│  │  (Model change nahi hota, sirf instruction)            │    │
│  └─────────────────────────────────────────────────────────┘    │
│                                                                  │
│  RAG:                                                            │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  User ──▶ Search ──▶ Documents ──▶ LLM ──▶ Output      │    │
│  │  (Model change nahi hota, external knowledge add)       │    │
│  └─────────────────────────────────────────────────────────┘    │
│                                                                  │
│  FINE-TUNING:                                                    │
│  ┌─────────────────────────────────────────────────────────┐    │
│  │  Data ──▶ Training ──▶ Model Weights Update ──▶ Output │    │
│  │  (Model ke andar change hota hai)                       │    │
│  └─────────────────────────────────────────────────────────┘    │
│                                                                  │
│  COMPARISON:                                                     │
│  ┌──────────────┬──────────────┬──────────────┐                │
│  │  Prompting   │     RAG      │ Fine-tuning  │                │
│  ├──────────────┼──────────────┼──────────────┤                │
│  │ Instruction  │ External     │ Weights      │                │
│  │              │ Knowledge    │ Update       │                │
│  ├──────────────┼──────────────┼──────────────┤                │
│  │ Fast         │ Medium       │ Slow         │                │
│  │ Cheap        │ Medium       │ Expensive    │                │
│  │ Limited      │ Good         │ Best         │                │
│  └──────────────┴──────────────┴──────────────┘                │
└─────────────────────────────────────────────────────────────────┘
```
